```python
import pandas as pd
import numpy as np

df=pd.read_csv("best-buy-products-updated.csv")
df = df[["sku","name", "customerReviewAverage", "customerReviewCount", "customerTopRated", "manufacturer", "class", "condition", "modelNumber",  "salePrice"]]

df.info()
```

```python
import json

with open(f"all-top10.json", 'r') as f:
    data = json.load(f)
top10 = pd.DataFrame(data)

top10.info()
```

```python
df["bestseller"] = df["sku"].astype(str).isin(top10["sku"].astype(str))
df["sku"] = df["sku"].astype(str)
top10["sku"] = top10["sku"].astype(str)

df = df.merge(top10[["sku", "rank"]], on="sku", how="left")

print(df["bestseller"].value_counts())
print(df["rank"].value_counts())
```

```python
df.to_csv("best-buy-products-bestseller-col.csv", index=False)
```
