```python
# Load dataset and view basic structure

import pandas as pd
import numpy as np

df=pd.read_csv("best-buy-products.csv")
df = df[["sku","name", "customerReviewAverage", "customerReviewCount", "customerTopRated", "manufacturer", "class", "condition", "modelNumber",  "salePrice"]]

# Display the first 5 rows for a quick preview
df.head()
```

```python
#remove rows that are empty
df = df.dropna()
```

```python
# Check structure
df.info()
```

```python
df.isna().sum()
```

```python
# Average Rating by Manufacturer

df.groupby("manufacturer")["customerReviewAverage"].mean().sort_values(ascending=False)
```

```python
# Statistical Checking — Pearson Correlation

# Define numeric_cols if it hasn't been defined yet (avoids overwriting if defined later)
if 'numeric_cols' not in globals():
	numeric_cols = ["salePrice", "customerReviewAverage", "customerReviewCount"]

df[numeric_cols].corr(method="pearson") # type: ignore
```

```python
# Price Distribution

from matplotlib import pyplot as plt


plt.figure(figsize=(8,4))
plt.hist(df['salePrice'], bins=30, color='skyblue', edgecolor='black')
plt.xlabel("Price ($)")
plt.ylabel("Number of Products")
plt.title("Distribution of Laptop Prices")
plt.show()
```

```python
# Rating Distribution

plt.figure(figsize=(8,4))
plt.hist(df['customerReviewAverage'], bins=20, color='lightgreen', edgecolor='black')
plt.xlabel("Average Rating")
plt.ylabel("Count")
plt.title("Distribution of Customer Ratings")
plt.show()
```

```python
# Average Rating Distribution

#plt.figure(figsize=(6,6))
plt.scatter(df['customerReviewCount'], df['customerReviewAverage'], alpha=0.5)
plt.xscale("log")
plt.xlabel("Number of Reviews (log)")
plt.ylabel("Average Rating")
plt.title("Rating vs Number of Reviews")
plt.show()

```

```python
# Price vs Rating (Scatter)

#plt.figure(figsize=(6,6))
plt.scatter(df['salePrice'], df['customerReviewAverage'], alpha=0.5, color='darkorange')
plt.xlabel("Price ($)")
plt.ylabel("Avg Rating")
plt.title("Price vs Rating")
plt.show()
```

```python
# Price vs Review Count (Scatter)

#plt.figure(figsize=(6,6))
plt.scatter(df['salePrice'], df['customerReviewCount'], alpha=0.5, color='darkorange')
plt.yscale("log")
plt.xlabel("Price ($)")
plt.ylabel("Number of Reviews")
plt.title("Price vs Number of Reviews")
plt.show()
```

```python
# Correlation Heatmap

import seaborn as sns

numeric_cols = ["salePrice", "customerReviewAverage", "customerReviewCount"]

plt.figure(figsize=(6,5))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Between Price, Ratings, and Reviews")
plt.show()
```

```python
# # Average Rating by Manufacturer (Bar Chart)

plt.figure(figsize=(10,4))
(df.groupby("manufacturer")["customerReviewAverage"]
   .mean()
   .sort_values()
   .plot(kind='bar', color='salmon'))
plt.ylabel("Average Rating")
plt.title("Average Rating by Manufacturer")
plt.show()


```

```python
# Reviews by Manufacturer (Bar Chart)

plt.figure(figsize=(10,4))
(df.groupby("manufacturer")["customerReviewCount"]
   .sum()
   .sort_values()
   .plot(kind='bar', color='salmon'))
plt.ylabel("Total Reviews (log scale)")
plt.yscale("log")
plt.title("Total Number of Reviews by Brand")
plt.show()

```

```python
# Top Rated Products
top_rated = df[df['customerTopRated'] == True]  
top_rated = top_rated.sort_values(by='customerReviewAverage', ascending=False).head(10)

# Best Sellers
best_sellers = df.sort_values(by='customerReviewCount', ascending=False).head(10)

print("Top Rated Products:")
print(top_rated[['name', 'customerReviewAverage', 'customerReviewCount', 'salePrice']]) 
```

```python
# Top 10 Highest Rated Products Visualization

plt.style.use('seaborn-v0_8-pastel')

top10 = df.sort_values(by='customerReviewAverage', ascending=False).head(10)

plt.figure(figsize=(10,6))
bars = plt.barh(top10['name'], top10['customerReviewAverage'])
plt.gca().invert_yaxis()
plt.xlabel("Average Rating ★")
plt.title("Top 10 Rated Laptops")

for bar in bars:
    plt.text(bar.get_width() + 0.01, bar.get_y() + bar.get_height()/2,
             f'{bar.get_width():.2f}', va='center', fontsize=10)

plt.show()
```

```python
# # Top 10 Highest Rated Products Visualization (per Manufacturer)

plt.style.use('seaborn-v0_8-pastel')

# Filter: require at least 10 reviews
df_filtered = df[df['customerReviewCount'] >= 10]

# Select best product from each manufacturer
top_brand_products = (
    df_filtered.sort_values(by='customerReviewAverage', ascending=False)
    .groupby('manufacturer')
    .first()
    .reset_index()
)

# Sort resulting list by rating DESCENDING
top_brand_products = top_brand_products.sort_values(by='customerReviewAverage', ascending=False)

plt.figure(figsize=(10,6))
bars = plt.barh(top_brand_products['name'], top_brand_products['customerReviewAverage'], color='seagreen')
plt.gca().invert_yaxis()
plt.xlabel("Average Rating ★")
plt.title("Top Rated Product from Each Manufacturer (Sorted by Rating)")

# Add rating labels
for bar in bars:
    width = bar.get_width()
    plt.text(width + 0.01, bar.get_y() + bar.get_height()/2,
             f'{width:.2f}', va='center', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
# Best Sellers Top 10 (Most Review) Visualization

plt.figure(figsize=(10,4))
plt.bar(best_sellers['name'][:10], best_sellers['customerReviewCount'][:10], color='skyblue')
plt.xticks(rotation=70, ha='right')
plt.ylabel("Number of Reviews")
plt.title("Top 10 Most Reviewed (Best Selling) Products")
plt.show()

```

```python
# Combined View: Ratings vs Reviews

plt.figure(figsize=(6,6))
plt.scatter(df['customerReviewCount'], df['customerReviewAverage'], alpha=0.5)
plt.xscale("log")
plt.xlabel("Number of Reviews (log scale)")
plt.ylabel("Average Rating")
plt.title("Top Rated vs Best Seller — Combined View")
plt.show()
```
