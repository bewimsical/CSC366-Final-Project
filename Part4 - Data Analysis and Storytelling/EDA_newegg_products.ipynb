## Descriptive Statistics

```python
# Summary statistics for numeric columns
df.describe()
df.describe(percentiles=[.1, .25, .5, .75, .9])
```

```python
# Summary statistics for categorical columns
df.describe(include='object')
```

```python
# Univariate Analysis — Numerical
print('Review count Summary:\n ', df['customerReviewCount'].describe())
print('\nReview avg Score Summary:\n ', df['customerReviewAverage'].describe())
print('\n Sale Price Summary:\n ', df['salePrice'].describe())

# Univariate Analysis — Categorical
print('\nManufacturer Summary:\n ', df['manufacturer'].describe())
print('\nclass Summary:\n ', df['class'].describe())
print('\nModel number Summary:\n ', df['modelNumber'].describe())
print('\ncondition Summary:\n ', df['isRefurbished'].describe())
print('\nManufacturer num:\n ', df['manufacturer'].value_counts())
print('\nclass num:\n ', df['class'].value_counts())
```

## Summary Statistics Review

```python

# Numerical summaries
print("Review Count")
print("Mean Review Count:", df['customerReviewCount'].mean())
print("Median Review Count:", df['customerReviewCount'].median())
print("Mode Review Count:", df['customerReviewCount'].mode()[0])
print("Std of Review Count:",df['customerReviewCount'].std() )
print("Variance of Review Count:",df['customerReviewCount'].var() )
print("Min & Max Review Count:",df['customerReviewCount'].min(), df['customerReviewCount'].max() )

# Using IQR (inter quartile range) method
Q1 = df['customerReviewCount'].quantile(.25)
Q3 = df['customerReviewCount'].quantile(.75)
IQR = Q3 - Q1

lower_fence = Q1 - IQR*1.5
upper_fence = Q3 + IQR*1.5
outliers = df[(df['customerReviewCount']<lower_fence)|(df['customerReviewCount']>upper_fence)]
print("Number of outliers Review Count:", outliers.shape[0])

# Using Z-Score method
from scipy import stats
z = stats.zscore(df['customerReviewCount'])
print("Potential outliers (Z > 3):", df[abs(z) > 3].shape[0])

print("\nReview Rating")
print("Mean Review Rating:", df['customerReviewAverage'].mean())
print("Median Review Rating:", df['customerReviewAverage'].median())
print("Mode Review Rating:", df['customerReviewAverage'].mode()[0])
print("Std of Review Rating:",df['customerReviewAverage'].std() )
print("Variance of Review Rating:",df['customerReviewAverage'].var() )
print("Min & Max Review Rating:",df['customerReviewAverage'].min(), df['customerReviewAverage'].max() )

# Using IQR (inter quartile range) method
Q1 = df['customerReviewAverage'].quantile(.25)
Q3 = df['customerReviewAverage'].quantile(.75)
IQR = Q3 - Q1

lower_fence = Q1 - IQR*1.5
upper_fence = Q3 + IQR*1.5
outliers = df[(df['customerReviewAverage']<lower_fence)|(df['customerReviewAverage']>upper_fence)]
print("Number of outliers Review Rating:", outliers.shape[0])

# Using Z-Score method
z = stats.zscore(df['customerReviewAverage'])
print("Potential outliers (Z > 3):", df[abs(z) > 3].shape[0])

print("\nSale Price")
print("Mean Sale Price:", df['salePrice'].mean())
print("Median Sale Price:", df['salePrice'].median())
print("Mode Sale Price:", df['salePrice'].mode()[0])
print("Std of Sale Price:",df['salePrice'].std() )
print("Variance of Sale Price:",df['salePrice'].var() )
print("Min & Max Sale Price:",df['salePrice'].min(), df['salePrice'].max() )

# Using IQR (inter quartile range) method
Q1 = df['salePrice'].quantile(.25)
Q3 = df['salePrice'].quantile(.75)
IQR = Q3 - Q1

lower_fence = Q1 - IQR*1.5
upper_fence = Q3 + IQR*1.5
outliers = df[(df['salePrice']<lower_fence)|(df['salePrice']>upper_fence)]
print("Number of outliers Sale Price:", outliers.shape[0])

# Using Z-Score method
z = stats.zscore(df['salePrice'])
print("Potential outliers (Z > 3):", df[abs(z) > 3].shape[0])

# Categorical summaries
print('\nManufacturer counts:\n ', df['manufacturer'].value_counts())
print('\nClass Counts:\n ', df['class'].value_counts())

```

## Exploring Relationships

```python
# Group-based summaries

print("Mean Rating by Class:")
print(df.groupby('class')['customerReviewAverage'].mean())

print("\nMean Rating by Manufacturer:")
print(df.groupby('manufacturer')['customerReviewAverage'].mean())

print("\nMean Review Count by Class:")
print(df.groupby('class')['customerReviewCount'].mean())

print("\nMean Review Count by Manufacturer:")
print(df.groupby('manufacturer')['customerReviewCount'].mean())

print("\nMean Sale Price by Class:")
print(df.groupby('class')['salePrice'].mean())

print("\nMean Sale Price by Manufacturer:")
print(df.groupby('manufacturer')['salePrice'].mean())


# Correlation matrix

df.corr(numeric_only=True)
```
